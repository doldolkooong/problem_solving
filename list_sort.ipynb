### 문제  

#### - 문제 예시 >> [8, 6, 2, 8, 5, 1, 2, 5, 4, 9]
#### - 최소값 : 1, 최대값 : 9, 중간값 5

1. 리스트에 있는 데이터들 정렬 -> sort() 
2. 가장 작은 수 찾기 -> [0]
3. 가장 큰 수 찾기 -> [-1]
4. 중간값 구하기
    - 1) 데이터 수가 홀수면 가운데값 찾기 -> %2 != 0
    - 2) 데이터 수가 짝수면 가운데 두개의 값 평균 내기

```python
# 버블 정렬 사용

lst = [8, 6, 2, 8, 5, 1, 2, 5, 4, 9] # 리스트 선언

temp = []

# 리스트 정렬 
for j in range(len(lst)-1):
    for i in range(len(lst)-1):

        if lst[i] > lst[i+1]:
            lst[i], lst[i+1] = lst[i+1], lst[i]

        elif lst[i] < lst[i+1]:
            lst[i], lst[i+1] = lst[i], lst[i+1]


print("정렬된 리스트 : ", lst)

# 최대값 구하는 방법
print("최대값 : ", lst[-1])

# 최소값 구하는 방법
print("최소값 : ", lst[0])

# 중간값 구하기
if len(lst) % 2 == 0: # 리스트 전체 길이가 짝수면 가운데 값 두 개의 평균
    print("중간값 : ", ((lst[len(lst)//2-1] + lst[len(lst)//2])/2))
else: # 리스트 전체 길이가
    print("중간값 : ", lst[len(lst//2)])
```

```python
# sort 함수 사용 
lst = [8, 6, 2, 8, 5, 1, 2, 5, 4, 9] # 리스트 선언

lst.sort() # sort 함수를 활용하여 정렬

min = lst[0] # 최소값 저장
max = lst[-1] # 최대값 저장

print("정렬된 리스트 : ", lst)
print("최대값 : ", max)
print("최소값 : ", min)

if len(lst) % 2 == 0 : # 길이가 짝수면 중간값 두 개 평균
    print("중간값 : ", ((lst[len(lst)//2] + lst[len(lst)//2 - 1])/2))
else: # 길이가 홀수면 중간값 하나 출력
    print("중간값 : ", lst[len(lst)//2])
```
