## [소수 구별](https://freedeveloper.tistory.com/391)
프로그래머 X는 입력값을 숫자를 입력하거나 문자를 입력하려고 하는데,
- 소수란 1보다 큰 자연수(양의 정수) 중에서 1과 자기 자신을 제외한 자연수로는 나누어떨어지지 않는 자연수이다.
- 만약 숫자를 입력하였으면 그것이 소수(2,3,5,7,11,13 ..)인지, 아닌지 구별하는 프로그램을 짜보도록 하고,
- 만약 문자를 입력하였으면 숫자가 아니므로 math error를 표시하게 하라.

1. 사용자의 입력값 받기 input
2. 소수 계산하는 식    
3. 숫자를 입력했다면(try-except)
    - if 숫자 % i == 0:
4. 글자를 입력했다면
    - math error 출력

```python
def is_prime(number):
    if number < 2:
        print("소수가 아닙니다.")
        return False
    
    
    for i in range(2, number): # 2부터 number까지 나머지 계산 해봄
        # number가 해당 수로 나누어떨어진다면 소수가 아님
        if number % i == 0:
            print(f"{number}는 소수가 아닙니다.")
            return False
    return True


try: # 숫자를 입력했을 경우
    number = int(input("숫자를 입력하시오 : "))

    if is_prime(number) is True:
        print(f"{number}는 소수입니다.")
    else:
        print(f"{number}는 소수가 아닙니다.")


except:# 글자를 입력했다면 math error 출력
    print("math error")
```
