## [방탈출 게임](https://codingdojang.com/scode/726#answer-filter-area)
- 비밀번호는 3자리로 한다.
- 비밀번호를 입력하게 한다.      
- 비밀번호가 틀리면 처음부터 다시 입력하게 한다.
- 비밀번호가 맞으면 성공 메시지를 출력한다.

1. 비빌번호 저장 변수
2. 사용자에게 비밀번호 입력 받기
3. 입력한 길이기 3이 맞는지 확인
4. 비밀번호 맞는지 확인
    - 맞으면 성공 메세지 출력
    - 틀리면 다시 입력
5. 몇 번 풀었는지 횟수 출력

```python
# 비밀번호 랜덤하게 받기
import random

secret = str(random.randint(100,999)) # 비밀번호 무작위
count = 0

while True:
    count += 1 # 한번 시도할 때마다 +1
    user = input("비밀번호를 입력하세요: ")

    if len(user) != 3:
        print("비밀번호는 3글자입니다. 다시 작성하세요")

    elif not user.isdigit():
        print("숫자를 입력하세요.")

    elif user != secret: # 틀리면 맞출 때까지 반복
        print(f"입력한 비밀번호 : {user} 틀렸습니다")
    else: # 정답을 맞추면 종료
        print("정답입니다!")
        print(f"총 시도 횟수 : {count}")
        break
```
