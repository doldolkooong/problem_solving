1. 처음에는 가위 바위 보로 시작한다.
2. 가위 바위 보 결과에 따라 공격 결정
   - user 승리 -> user 공격
   - computer 승리 -> computer 공격
3. 공격권을 가진 사람이 가위 바위 보 진행
4. 공격권을 가진 사람이 이기면 공격권 유지
5. 공격권을 가진 사람이 지면 공격권을 상대에게 넘김
6. 처음 이후 가위바위보 결과가 무승부이면 공격권 가진 사람 승리 
7. 승자가 결정되면 게임 종료

```python
import random
import enum

class rsp_choice(enum.Enum): # 가위, 바위, 보에 값을 넣은 상수 생성
    가위 = 0
    바위 = 1
    보 = 2

def mukchippa():
    attacker = None

    # 1. 누가 공격할 건지 정하기 - 가위바위보 반복문
    while True:
        computer_choice = random.choice(list(rsp_choice.__members__)) # rsp_choice에서 computer가 랜덤으로 선택
        user_choice = input("가위, 바위, 보 중 하나 적어주세요: ") # 사용자의 입력 받음

        if user_choice not in rsp_choice.__members__: # 사용자가 rsp_choice에서 선택하지 않은 경우 출력
                print("잘못 입력했습니다. 다시 입력해주세요.")
                continue
        
        print(f"User: {user_choice}") # 사용자의 선택 출력
        print(f"Computer: {computer_choice}") # 컴퓨터의 선택 출력

        # 같은 것을 냈으면 게임 진행이 안됨
        # 다시 고를 수 있도록
        if user_choice == computer_choice:
            print("====비겼습니다. 다시 가위바위보!====")
            continue

        # 가위 바위 보 계산
        result = (rsp_choice[user_choice].value - rsp_choice[computer_choice].value) % 3

        # User 승리 -> User의 공격
        if result == 1: 
            attacker = "User"
            print("User가 공격권을 가져갑니다.")
            break

        # Computer 승리 → Computer의 공격
        elif result == 2:
            attacker = "Computer"
            print("====Computer가 공격권을 가져갑니다.====")
            break


    # 2. 묵찌빠 진행 반복문
    while True:
        print(f"<현재 공격권 : {attacker}>") # 공격 할 사람 저장

        user_choice = input("가위, 바위, 보 중 하나 적어주세요: ") # 사용자의 입력 받음
        computer_choice = random.choice(list(rsp_choice.__members__))

        if user_choice not in rsp_choice.__members__: # user가 rsp_choice에 없는 것을 입력할 경우
            print("잘못 입력했습니다. 다시 입력해주세요.")
            continue

        print(f"User: {user_choice}")
        print(f"Computer: {computer_choice}")

        result = (rsp_choice[user_choice].value - rsp_choice[computer_choice].value) % 3

        if result == 1: # user의 승리 -> 공격권은 user에게
            attacker = "User" # 공격권에 User 저장
            print("====User가 공격권을 가져갑니다.====")

        elif result == 2:  # computer의 승리 -> 공격권은 computer에게
            attacker = "Computer" # 공격권에 Computer 저장
            print("====Computer가 공격권을 가져갑니다.====")

        else: # 만약 같은 것을 내면 attacker의 승리
            print(f"~~{attacker}의 승리~~")
            break


mukchippa()
```
