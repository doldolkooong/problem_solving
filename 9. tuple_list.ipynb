## 문제 
- 원점(0,0)으로부터의 거리가 3 이하인 점들만 필터링
    - 거리 공식: sqrt(x^2 + y^2)

````python
import math

points = [(1, 2), (3, 4), (-1, 5), (2, -3), (0, 0), (4, 1), (-2, -2)]

x = [points[i][0] for i in range(len(points))] 
y = [points[i][1] for i in range(len(points))]


for i in range(len(points)):
    distance[i] = math.sqrt(x[i]**2 + y[i]**2)

    if distance[i] <= 3:
        print(points[i])
````

### 문제 2: 모든 점을 x축 기준으로 대칭 이동시킨 새로운 좌표 리스트 생성

```python
## 문제 
- 모든 점을 x축 기준으로 대칭 이동시킨 새로운 좌표 리스트 생성
  - x축 대칭: (x, y) -> (x, -y)
```

````python
import math

points = [(1, 2), (3, 4), (-1, 5), (2, -3), (0, 0), (4, 1), (-2, -2)]

lst_points=[]

for i in range(len(points)):
    lst_points.append(list(points[i])) # points 안에 있는 튜플들을 리스트로 변경

    x = lst_points[i][0] 

    if x > 0:
        lst_points[i][0] = -x # x값이 0보다 크면 -x로 변경
    else:
        lst_points[i][0] = abs(x) # x값이 0보다 작으면 +x로 변경

final_points = []
for i in range(len(points)): # 리스트로 바꾼 튜플들을 다시 튜플로 변경
    final_points.append(tuple(lst_points[i]))
    

final_points
````

문제 3: 1사분면(x > 0, y > 0)에 있는 점들의 개수 계산

````python
import math

points = [(1, 2), (3, 4), (-1, 5), (2, -3), (0, 0), (4, 1), (-2, -2)]

lst_points=[]
count = 0

for i in range(len(points)):
    lst_points.append(list(points[i])) # points 안에 있는 튜플들을 리스트로 변경

    x = lst_points[i][0] # ex. (1, 2) 중 첫번째 값
    y = lst_points[i][1] # ex. (1, 2) 중 두번째 값

    if x > 0 and y > 0: # 1사분면에 있는 경우
        count += 1 # count가 1씩 올라감

count
````

문제 4: 각 점의 원점으로부터의 거리를 계산한 리스트 생성

````python
import math

points = [(1, 2), (3, 4), (-1, 5), (2, -3), (0, 0), (4, 1), (-2, -2)]

lst_points=[]
origin_distance = []

count = 0

for i in range(len(points)):
    lst_points.append(list(points[i])) # points 안에 있는 튜플들을 리스트로 변경

    x = lst_points[i][0] # ex. (1, 2) 중 첫번째 값
    y = lst_points[i][1] # ex. (1, 2) 중 두번째 값

    origin_distance.append(math.sqrt(x**2 + y**2)) # 원점으로부터의 거리 계산 후 리스트에 저장

    print(f"{origin_distance[i]:.2f}")

````
