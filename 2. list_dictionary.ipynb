문제 1: 각 학생의 평균 점수를 계산하여 딕셔너리로 반환하는 함수

1. 점수 딕셔너리만 따로 저장하는 리스트 하나 생성
2. 점수 리스트에서 합 세는 함수 생성
3. 학생 이름과 점수 출력

````python
students = [
    {"name": "최민준", "subjects": {"수학": 76, "영어": 84, "과학": 91}},
    {"name": "한서연", "subjects": {"수학": 95, "영어": 79, "과학": 88}},
    {"name": "강도윤", "subjects": {"수학": 83, "영어": 87, "과학": 92}},
    {"name": "윤하은", "subjects": {"수학": 68, "영어": 93, "과학": 75}}
]
subject = []

for i in range(len(students)):
    subject.append(students[i]["subjects"])

def score_total(subject : list):
    dic_total = {}

    for i in range(len(subject)):
        total = sum(subject[i].values())//3
        dic_total[students[i]["name"]] = total

        #print(f"{students[i]["name"]} : {total[i]}")

    return dic_total

score_total(subject)
````

문제 2: 모든 과목에서 80점 이상인 학생들의 이름을 리스트로 반환하는 함수

````python
students = [
    {"name": "최민준", "subjects": {"수학": 76, "영어": 84, "과학": 91}},
    {"name": "한서연", "subjects": {"수학": 95, "영어": 79, "과학": 88}},
    {"name": "강도윤", "subjects": {"수학": 83, "영어": 87, "과학": 92}},
    {"name": "윤하은", "subjects": {"수학": 68, "영어": 93, "과학": 75}}
]
subject = []

for i in range(len(students)): # 과목 점수들만 리스트에 저장
    subject.append(students[i]["subjects"])


def all_eighty_scores(subject : list):
    name = []
    for i in range(len(subject)):
        if subject[i]["수학"] >= 80 and subject[i]["영어"] >= 80 and subject[i]["과학"] >= 80: # 모든 과목 80점 넘으면
            name.append(students[i]["name"]) # 해당하는 이름 출력

    return name

all_eighty_scores(subject)
````

문제 3: 가장 높은 평균 점수를 가진 학생을 찾는 함수

````python
students = [
    {"name": "최민준", "subjects": {"수학": 76, "영어": 84, "과학": 91}},
    {"name": "한서연", "subjects": {"수학": 95, "영어": 79, "과학": 88}},
    {"name": "강도윤", "subjects": {"수학": 83, "영어": 87, "과학": 92}},
    {"name": "윤하은", "subjects": {"수학": 68, "영어": 93, "과학": 75}}
]
subject = []

for i in range(len(students)): # 과목 점수들만 리스트에 저장
    subject.append(students[i]["subjects"])

def max_score(subject : list): # 평균이 가장 높은 학생 구하는 함수
    dic_total = {} # 딕셔너리

    for i in range(len(subject)): # 학생 이름과 평균을 딕셔너리로 합치는 반복문
        total = sum(subject[i].values())//3
        dic_total[students[i]["name"]] = total

    best_student = max(dic_total.items(), key=lambda x:x[1]) # 평균 가장 높은 학생 구하기

    best_student = {
        best_student[0] : best_student[1] # 딕셔너리로 변경
    }

    return best_student

max_score(subject)
````
