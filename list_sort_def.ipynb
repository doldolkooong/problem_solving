< 문제 > 함수로 구현해주세요.
예시 >> [8, 6, 2, 8, 5, 1, 2, 5, 4, 9]
최소값: 1 / 최대값: 9 / 중간값: 5
1. 리스트 정렬 
2. 최소값 구하는 함수
3. 최대값 구하는 함수
4. 중간값 구하는 함수


```python
# 리스트 정렬
lst = [8, 6, 2, 8, 5, 1, 2, 5, 4, 9]

def lst_sort():
    lst.sort()
    return lst

def lst_min(): # 리스트 최소값구하는 함수
    min = lst[0]
    
    return min

def lst_max(): # 리스트 최대값 구하는 함수
    max = lst[-1]

    return max

def lst_med(): # 중간값 구하는 함수
    if len(lst) % 2 == 0:
        med = ((lst[len(lst)//2-1] + lst[len(lst)//2])/2)
    elif len(lst) % 2 != 0:
        med = lst[len(lst)//2]

    return med

print("정렬된 리스트 : ", lst_sort())
print("최소값 : ", lst_min())
print("최대값 : ", lst_max())
print("중간값 : ", lst_med())
```
