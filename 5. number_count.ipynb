## [숫자 세기 알고리즘](https://codingdojang.com/scode/720#answer-filter-area)
- 0부터 9999까지 8을 포함하지 않은 수는 총 몇개일까?
- 8, 108, 888, 998 등은 8을 포함하는 수입니다.
- 111, 299, 437 등은 8을 포함하지 않은 수입니다.

1. 숫자를 문자열로 변환
2. 8이 아닌 수 저장할 변수
3. 8이 몇개 있는지 개수 확인
    - 있으면 변함 없음
    - 없으면 count +1    
4. 개수 출력

```python
count = 0 # 8이 아닌 개수 저장

for i in range(10000): # 0부터 10000-1까지
    if "8" not in str(i): # i 안에 8이 없을 경우
        count += 1 
    
print(count)
```

```python
count = 0 # 8이 아닌 개수 저장

for i in range(10000): # 0부터 10000-1까지
    number = str(i) # int -> str 변경

    if "8" in number: # 8이 있을 경우
        continue # for문 진행
    count += 1 # 8이 없을 경우 count + 1

print(f"8이 포함되지 않은 숫자 개수: {count}")
```
