## [Up&Down 숫자 맞추기](https://codingdojang.com/scode/711#answer-filter-area)
- 컴퓨터가 1 ~ 100 중 랜덤으로 숫자를 정합니다.
- 입력한 숫자보다 정답이 크면, "UP" 출력,
- 입력한 숫자보다 정답이 작으면, "DOWN" 출력,
- 정답을 맞추면, "정답" 출력
   
1. 컴퓨터의 선택(random.randint(1, 100))
2. 사용자의 입력값 받기
3. 숫자 맞추는 함수 생성
4. 입력한 숫자가 정답보다 크면 UP 출력
5. 입력한 숫자가 정답보다 작으면 Down 출력
6. 정답을 맞추면 정답 출력

```python
import random

computer_choice = random.randint(1,100) # 컴퓨터가 선택한 수


def number_correct(computer_choice):
    count = 0 # 맞추기 위해 시도한 횟수


    while True:
        user_choice = int(input("숫자를 입력하세요")) # 사용자 입력값 받기
        count += 1 # while문 한번 돌아갈 때마다 1 증가

        if user_choice > computer_choice: # 사용자의 선택이 정답보다 클 경우
            print("DOWN!!")
            print("======") # 사용자의 선택이 정답보다 작을 경우
        elif user_choice < computer_choice:
            print("UP !")
            print("======")
        else: # 정답을 맞추면 게임이 끝남
            print(f"맞추기 위해 시도한 횟수입니다 : {count}")
            print("정답!")
            break

number_correct(computer_choice)
```
